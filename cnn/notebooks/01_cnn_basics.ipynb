# Small PyTorch CNN Practice

This notebook aims to build a simple CNN and understand how image dimensions change as they pass through each layer. 

We will not be training a model in this notebook. We are only inspecting the architecture.

## 01. Create a set of dummy images

```python
import torch
import torch.nn as nn

# Batch of 8 64x64 grayscale (1 channel) images
x = torch.randn(8, 1, 64, 64)
print(x.shape)
```

## 02. Build a small CNN

```python
model = nn.Sequential(
    nn.Conv2d(in_channels=1,
        out_channels=16,
        kernel_size=3,
        stride=1,
        padding=1),
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2),

    nn.Conv2d(in_channels=16,
        out_channels=32, 
        kernel_size=3,
        padding=1),
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2),
    nn.Flatten(),
    nn.Linear(32 * 16 * 16, 2))
```

## 03. Inspect output shape at every layer

```python
x = torch.randn(8, 1, 64, 64)

for layer in model:
    x = layer(x)
    print(layer.__class__.__name__, x.shape)
```

Input image batch: 8x1x64x64
Convolution, 16 feature maps: 8x16x64x64
MP, half spatial dimensions: 8x16x32x32
Convolution, 32 feature maps: 8x32x32x32
MP, half spatial dimensions: 8x32x16x16
Flatten: 8x8192
Linear, 2 class scores per image: 8x2
