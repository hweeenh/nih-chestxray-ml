# Small ResNet Practice

This notebook aims to build a simple ResNet and understand how residual blocks and shortcut connections are used to optimise the addition of each layer, and to visualise the changes as they pass through each layer. 

We will not be training a model in this notebook. We are only inspecting the architecture.

```python
import torchvision.models as models

model = models.resnet18(
    weights=models.ResNet18_Weights.DEFAULT
)

print(model)
```

```python
# Components of the ResNet model
print("Conv1:",model.conv1)
print("BatchNorm1:",model.bn1)
print("ReLU:",model.relu)
print("MaxPool:",model.maxpool)

print("Layer1:",model.layer1)
print("Layer2:",model.layer2)
print("Layer3:",model.layer3)
print("Layer4:",model.layer4)

print("AvgPool:",model.avgpool)
print("FC:",model.fc)
```

```python
# Inspect projection shortcuts directly
print(model.layer2[0].downsample)
print(model.layer3[0].downsample)
print(model.layer4[0].downsample)
```
